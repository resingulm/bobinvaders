Hier ist ein vollständiges Beispiel für ein Jupyter Notebook, das ein klassisches Optimierungsproblem löst: das Rucksackproblem (Knapsack Problem).

Stell dir vor, du bist ein Dieb in einem Museum. Du hast einen Rucksack, der maximal 15 kg tragen kann. Es gibt verschiedene Ausstellungsstücke mit unterschiedlichen Werten und Gewichten. Welche solltest du mitnehmen, um den Gesamtwert zu maximieren?

```python
# 1. Installation der Gurobi-Bibliothek (für Google Colab notwendig)
!pip install gurobipy

import gurobipy as gp
from gurobipy import GRB

# 2. Daten des Problems
# Gegenstände: (Name, Wert, Gewicht)
items = {
    "Goldstatue": (2000, 10),
    "Gemälde": (1500, 5),
    "Vase": (800, 3),
    "Münzsammlung": (500, 2),
    "Skulptur": (1200, 8)
}

max_weight = 15

# 3. Modell erstellen
model = gp.Model("RucksackProblem")

# 4. Entscheidungsvariablen hinzufügen
# Für jeden Gegenstand erstellen wir eine Binärvariable (0 oder 1)
# 1 = Einpacken, 0 = Da lassen
x = {}
for item in items:
    x[item] = model.addVar(vtype=GRB.BINARY, name=item)

# 5. Zielfunktion festlegen: Maximiere den Gesamtwert
model.setObjective(
    gp.quicksum(x[item] * items[item][0] for item in items),
    GRB.MAXIMIZE
)

# 6. Nebenbedingung hinzufügen: Gewichtsbeschränkung
model.addConstr(
    gp.quicksum(x[item] * items[item][1] for item in items) <= max_weight,
    "Gewichtslimit"
)

# 7. Optimierung starten
model.optimize()

# 8. Ergebnis ausgeben
if model.status == GRB.OPTIMAL:
    print("\n--- Optimale Lösung gefunden ---")
    total_value = 0
    total_weight = 0
    for item in items:
        if x[item].X > 0.5:  # Wenn die Variable 1 ist
            val, weight = items[item]
            print(f"Eingepackt: {item} (Wert: {val}€, Gewicht: {weight}kg)")
            total_value += val
            total_weight += weight

    print(f"\nGesamtwert: {total_value}€")
    print(f"Gesamtgewicht: {total_weight}kg")
else:
    print("Keine optimale Lösung gefunden.")
```
